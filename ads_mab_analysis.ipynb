# Multi-Armed Bandit Analysis for Advertisement CTR

This notebook loads the advertisement click-through-rate dataset, computes CTR per ad, and runs an epsilon-greedy multi-armed bandit simulation to identify the best ad.

```python
import pandas as pd
import numpy as np
from pathlib import Path

path = Path('Ads_CTR_Optimisation.csv')
df = pd.read_csv(path)
print('Dataset shape:', df.shape)
print(df.head())
```

```python
ctr = df.mean().sort_values(ascending=False)
print('CTR by ad (in %):')
print((ctr * 100).round(2))
```

```python
num_ads = df.shape[1]
rewards = np.zeros(num_ads)
counts = np.zeros(num_ads)
epsilon = 0.1

for i in range(len(df)):
    if np.random.rand() < epsilon:
        chosen_ad = np.random.randint(num_ads)
    else:
        avg_reward = np.divide(rewards, counts, out=np.zeros(num_ads), where=counts > 0)
        chosen_ad = np.argmax(avg_reward)

    reward = df.iloc[i, chosen_ad]
    rewards[chosen_ad] += reward
    counts[chosen_ad] += 1

final_avg_reward = np.divide(rewards, counts, out=np.zeros(num_ads), where=counts > 0)
best_ad = int(np.argmax(final_avg_reward))

print('Best ad selected by epsilon-greedy bandit:', best_ad)
print('Average reward per ad:', final_avg_reward)
print('Best ad CTR from dataset:', round(ctr.iloc[best_ad] * 100, 2), '%')
```
