```python
# CartPole MDP Exploration in Jupyter Notebook

# Import necessary libraries
import gym
import numpy as np
np.bool8=np.bool_

# Load the CartPole environment
env = gym.make('CartPole-v1')

# Reset the environment to get the initial state
state = env.reset()

# Display the action space and observation space
print(f"Action Space: {env.action_space}")
print(f"Observation Space: {env.observation_space}")

# Define MDP Components for CartPole

# State space
def describe_state(state):
    """
    This function prints out the individual components of the state
    State is a tuple (x, x_dot, theta, theta_dot)
    """
    cart_position, cart_velocity, pole_angle, pole_velocity = state
    print(f"Cart Position: {cart_position}")
    print(f"Cart Velocity: {cart_velocity}")
    print(f"Pole Angle: {pole_angle}")
    print(f"Pole Velocity: {pole_velocity}")

# Example of an initial state in CartPole
print("Initial State:")
describe_state(state)

# Action space exploration
# In CartPole, there are two actions: 0 (push left) and 1 (push right)
actions = {0: "Move Left", 1: "Move Right"}
for action in actions:
    print(f"Action {action}: {actions[action]}")

# Simulate a few steps to see state transitions and rewards
num_steps = 5
print("\nSimulating a few steps:")
for step in range(num_steps):
    action = env.action_space.sample()  # Random action
    next_state, reward, done, info = env.step(action)

    print(f"\nStep {step + 1}:")
    print(f"Action taken: {actions[action]}")
    print("Next State:")
    describe_state(next_state)
    print(f"Reward: {reward}")
    print(f"Done: {done}")

# Close the environment when done
env.close()

```

## Taxi-v3 Environment

```python
import gym

# Load the Taxi environment
env_taxi = gym.make('Taxi-v3')

# Display the action space and observation space
print(f"Action Space: {env_taxi.action_space}")
print(f"Observation Space: {env_taxi.observation_space}")

# MDP Components for Taxi

# State space: Discrete(500) representing taxi's row, column, passenger location, and destination.
# state = (taxi_row, taxi_col, passenger_location, destination)
print(f"\nState Space: Discrete({env_taxi.observation_space.n}) states. Each state encodes (taxi_row, taxi_col, passenger_location, destination).")

# Action space: Discrete(6) - 0:South, 1:North, 2:East, 3:West, 4:Pickup, 5:Dropoff
actions_taxi = {0: "South", 1: "North", 2: "East", 3: "West", 4: "Pickup", 5: "Dropoff"}
print(f"Action Space: Discrete({env_taxi.action_space.n}) actions.")
for action, desc in actions_taxi.items():
    print(f"  Action {action}: {desc}")

# Reward: -1 for each regular time step, -10 for illegal pickup/dropoff, +20 for a successful dropoff.
print("\nReward: -1 per step, -10 for illegal pickup/dropoff, +20 for successful dropoff.")

# Transition: Deterministic movements for valid actions. Picking up/dropping off changes passenger/destination state.
print("Transition Dynamics: Moving actions (South, North, East, West) are deterministic unless hitting a wall. Pickup/Dropoff actions change the state based on passenger/destination presence.")

# Simulate a few steps
print("\nSimulating a few steps in Taxi:")
state_taxi = env_taxi.reset() # Removed [0] as env.reset() returns the state directly in old API
print(f"Initial State: {state_taxi}")
for step in range(3):
    action_taxi = env_taxi.action_space.sample()  # Random action
    # Adjusted unpacking for old Gym API which returns 4 values (observation, reward, done, info)
    next_state_taxi, reward_taxi, done_taxi, info_taxi = env_taxi.step(action_taxi)

    print(f"\nStep {step + 1}:")
    print(f"  Action taken: {actions_taxi[action_taxi]}")
    print(f"  Next State: {next_state_taxi}")
    print(f"  Reward: {reward_taxi}")
    print(f"  Done: {done_taxi}")
    if done_taxi:
        print("  Episode terminated.")
        break
env_taxi.close()
```
