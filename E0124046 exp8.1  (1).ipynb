```python
from collections import defaultdict
import gymnasium as gym
import numpy as np

def random_policy(state):
    return np.random.choice([0, 1, 2, 3])
def every_visit_mc_prediction(env, policy, num_episodes, gamma=0.95):

    returns_sum = defaultdict(float)
    returns_count = defaultdict(float)

    V = defaultdict(float)

    for episode_num in range(1, num_episodes + 1):
        if episode_num % 5000 == 0:
            print(f"Simulating Episode {episode_num}/{num_episodes}...")
        episode = []
        state, info = env.reset()
        done = False

        while not done:
            action = policy(state)
            next_state, reward, terminated, truncated, info = env.step(action)
            done = terminated or truncated
            episode.append((state, action, reward))
            state = next_state

        G = 0.0

        for i in range(len(episode) - 1, -1, -1):
            s, a, r = episode[i]
            G = gamma * G + r

            returns_sum[s] += G
            returns_count[s] += 1
            V[s] = returns_sum[s] / returns_count[s]

    return V
env = gym.make('FrozenLake-v1', is_slippery=False)

num_episodes = 20000
state_values = every_visit_mc_prediction(env, random_policy, num_episodes)
env.close()
print("\n--- Final 4x4 Grid State Value Estimates ---")
value_grid = np.zeros((4, 4))
for state, value in state_values.items():
    row, col = divmod(state, 4)
    value_grid[row, col] = value

print(np.round(value_grid, 4))
```
